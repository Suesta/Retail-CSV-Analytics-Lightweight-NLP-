# Ejercicio 1

Escribe una función `my_gcd(a, b, verbose=False)` que calcule el máximo común divisor de dos números enteros positivos utilizando un algoritmo propio (por ejemplo, el algoritmo de Euclides).

La función debe devolver el MCD.

Si `verbose=True`, debe imprimir los pasos que sigue para encontrarlo.

Una vez implementada la función, compara el resultado con la función incorporada en Python `math.gcd(a, b)` para verificar que es correcto.

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px;">SM</span> **(1 punto)**

```python
def my_gcd(a: int, b: int, verbose: bool = False) -> int:
    """
    calculo el máximo común divisor (mcd) de dos enteros positivos con euclides (iterativo)

    parámetros

    a : int  → primer entero > 0
    b : int  → segundo entero > 0
    verbose : bool  → si True, voy mostrando los pasos

    returns

    int → mcd(a, b)

    notas

    me apoyo en que: gcd(a, b) = gcd(b, a % b) y repito hasta que el resto sea 0
    """
    # valido que me llegan enteros positivos
    if not (isinstance(a, int) and isinstance(b, int)):
        raise TypeError("a y b deben ser int.")
    if a <= 0 or b <= 0:
        raise ValueError("a y b deben ser > 0.")

    # copio a variables locales
    x, y = a, b

    # si piden verbose, voy imprimiendo cada paso
    if verbose:
        print(f"Iniciando Euclides para a={a}, b={b}")
    while y != 0:
        if verbose:
            q = x // y
            r = x % y
            print(f"{x} = {y} * {q} + {r}")
        x, y = y, x % y

    # ya tengo el resultado en x
    if verbose:
        print(f"MCD({a}, {b}) = {x}")
    return x


# ahora hago unas pruebas rápidas comparando con math.gcd
pares_prueba: List[Tuple[int, int]] = [
    (1, 1),
    (12, 18),
    (1071, 462),
    (100, 25),
    (81, 57),
    (270, 192),
]

print("Comparación con math.gcd:")
for a, b in pares_prueba:
    m1 = my_gcd(a, b, verbose=False)
    m2 = math.gcd(a, b)
    estado = "OK" if m1 == m2 else "MAL"
    print(f"  gcd({a}, {b}) -> my_gcd={m1}, math.gcd={m2}  {estado}")

# dejo un ejemplo con verbose=True para ver el desarrollo paso a paso
print("\nEjemplo verbose=True:")
_ = my_gcd(1071, 462, verbose=True)
```

# Ejercicio 3

Nos piden también crear una función que recorra todo el archivo `botiga_en_linia.csv`.

La función, llamada `create_dictionary`, recibirá como *entrada*:

- La ruta que contiene el archivo de datos en formato *.csv*.  
- El código de un cliente (`CustomerId`).

La *salida* será un diccionario donde:

- Cada clave debe contener el país.  
- El valor correspondiente a cada clave será una lista de tuplas.  
  La lista debe contener todos los productos vendidos a ese cliente y país (total de productos).

**Nota importante:**  
Utiliza los principios de **programación funcional** que hemos visto en teoría para resolver este ejercicio.

<span style="font-family: Courier New; background-color: #ffcc5c; color: #000000; padding: 3px;">EG</span> **(1 punto)**


```python
from functools import reduce

def create_dictionary(ruta_csv: str, customer_id: str) -> Dict[str, List[Tuple[str, int]]]:
    """
    construyo un diccionario por país con los productos comprados por un cliente
    y el total de unidades por producto (suma de Quantity) en ese país.

    Parámetros

    ruta_csv : str
        ruta al archivo 'botiga_en_linia.csv' (separador ';', coma decimal ',').
    customer_id : str
        código del cliente.

    Returns

    Dict[str, List[Tuple[str, int]]]
        diccionario con clave = país y valor = lista de tuplas (producto, total_unidades).

    Notas
    - uso map/filter/reduce para montar la estructura final.
    - leo con varios encodings por si hay tildes.
    """
    # leo el csv
    encodings = ["utf-8", "utf-8-sig", "latin-1", "cp1252"]
    ultimo_error = None
    df = None
    for enc in encodings:
        try:
            df = pd.read_csv(
                ruta_csv,
                sep=";",
                decimal=",",
                parse_dates=["Date"],
                dayfirst=False,
                encoding=enc,
                engine="python",
            )
            break
        except Exception as e:
            ultimo_error = e
    if df is None:
        # si todo falla, fuerzo lectura con reemplazo de caracteres
        df = pd.read_csv(
            ruta_csv,
            sep=";",
            decimal=",",
            parse_dates=["Date"],
            dayfirst=False,
            encoding="latin-1",
            errors="replace",
            engine="python",
        )

    # localizo las columnas por nombre
    cols = {c.lower(): c for c in df.columns}
    col_customer = cols.get("customerid") or cols.get("customer_id") or "CustomerID"
    col_country  = cols.get("country") or "Country"
    col_product  = cols.get("product") or "Product"
    col_qty      = cols.get("quantity") or "Quantity"

    # filtro solo las filas del cliente (ignoro mayúsculas)
    df_cli = df[df[col_customer].astype(str).str.lower() == customer_id.lower()].copy()

    # agrego total de unidades por (país, producto) para ese cliente
    df_grp = (
        df_cli.groupby([col_country, col_product], dropna=False)[col_qty]
              .sum(min_count=1)
              .fillna(0)
              .astype(int)
              .reset_index()
    )

    # convierto a tuplas (country, product, qty) y me quedo con qty>0
    tuplas = map(lambda r: (r[0], r[1], int(r[2])), df_grp[[col_country, col_product, col_qty]].itertuples(index=False, name=None))
    tuplas = filter(lambda t: t[2] > 0, tuplas)

    # reduzco a { country: [(product, qty), ..] }
    def _acum(dic: Dict[str, List[Tuple[str, int]]], t: Tuple[str, str, int]):
        pais, prod, qty = t
        if pais not in dic:
            dic[pais] = []
        dic[pais].append((prod, qty))
        return dic

    resultado: Dict[str, List[Tuple[str, int]]] = reduce(_acum, tuplas, {})

    # ordeno las listas por producto
    for pais in resultado:
        resultado[pais] = sorted(resultado[pais], key=lambda x: (str(x[0]).lower(), -int(x[1])))

    return resultado

# reutilizo la ruta detectada en el ej. 2 si existe
ruta_demo = ruta_csv if 'ruta_csv' in globals() and os.path.exists(ruta_csv) else \
            "/content/drive/MyDrive/Programación para la ciencia de datos/Colab Notebooks/Activity_2/data/botiga_en_linia.csv"

# cojo un customer_id válido
try:
    cliente_demo = str(df_tienda["CustomerID"].iloc[0])
except Exception:
    tmp_df = pd.read_csv(ruta_demo, sep=";", decimal=",", encoding="latin-1", engine="python")
    cliente_demo = str(tmp_df["CustomerID"].iloc[0])

print(f"— 3: demo de create_dictionary para CustomerID={cliente_demo} —")
dic_res = create_dictionary(ruta_demo, cliente_demo)

# imprimo el diccionario en formato legible
for pais, lista_prod in dic_res.items():
    print(f"{pais}:")
    for prod, qty in lista_prod[:8]:
        print(f"  - {prod}: {qty}")
    if len(lista_prod) > 8:
        print(f"  ... ({len(lista_prod)} productos en total)")
```

# Ejercicio 4

A continuación, se quiere crear una función que devuelva un nuevo diccionario que contendrá, como clave, todos los productos que contengan la combinación de palabras "ni" o "te" o que empiecen por la letra "u".  
Para cada producto encontrado, el valor del diccionario será el precio unitario.

<span style="font-family: Courier New; background-color: #ffcc5c; color: #000000; padding: 3px;">EG</span> **(1 punto)**


```python
# aquí quiero: clave = nombre de producto | valor = precio unitario y si hay varias filas, uso el promedio
from typing import Dict
import pandas as pd

def productos_diccionario_regex(df: pd.DataFrame) -> Dict[str, float]:
    """
    devuelvo un diccionario {producto: precio_unitario} con los productos que:
      - contienen la subcadena "ni", o
      - contienen la subcadena "te", o
      - empiezan por la letra "u"
    todo sin distinguir mayúsculas/minúsculas.

    si un producto aparece en varias filas, uso el precio unitario medio (UnitPrice).
    """
    # mapeo nombres de columnas a minúsculas para localizar las que necesito
    cols = {c.lower(): c for c in df.columns}
    col_prod = cols.get("product", "Product")
    col_price = cols.get("unitprice", "UnitPrice")

    # calculo el precio medio por producto (por si un mismo producto sale varias veces)
    precios_medios = (
        df.groupby(col_prod, dropna=False)[col_price]
          .mean()
          .dropna()
    )

    # preparo el predicado de filtrado en minúsculas
    def _match(nombre: str) -> bool:
        n = str(nombre).strip().lower()
        return ("ni" in n) or ("te" in n) or n.startswith("u")

    # paso a pares (producto, precio) y filtro con el predicado anterior
    pares = ((prod, float(precio)) for prod, precio in precios_medios.items())
    pares_filtrados = ((p, v) for (p, v) in pares if _match(p))

    # construyo el diccionario final y redondeo a 2 decimales
    dicc = {p: round(v, 2) for p, v in pares_filtrados}
    return dicc

# llamo a la función con el df_tienda del ejercicio 2
dic_prod_precios = productos_diccionario_regex(df_tienda)

print("— 4: tamaño del diccionario de productos filtrados —")
print(f"nº de productos que cumplen: {len(dic_prod_precios)}")

print("\n— 4: primeras 10 entradas —")
cont = 0
for k, v in dic_prod_precios.items():
    print(f"  {k}: {v:.2f}")
    cont += 1
    if cont >= 10:
        break
```

```python
# aquí voy a tirar de zipfile y os.path.getsize
import os
import zipfile
from typing import Tuple

# monto Drive solo si estoy en Colab
try:
    import google.colab
    from google.colab import drive
    if not os.path.ismount("/content/drive"):
        drive.mount("/content/drive", force_remount=False)
except Exception:
    pass

def zip_decompression(ruta_zip: str, carpeta_destino: str) -> None:
    """
    descomprimo un archivo .zip en la carpeta destino.

    Parámetros

    ruta_zip : str
        ruta completa al archivo .zip (p. ej. 'Activity_2/data/books.zip').
    carpeta_destino : str
        ruta de la carpeta donde quiero extraer (si no existe, la creo).

    Returns

    None
        no devuelvo nada; dejo los ficheros extraídos en 'carpeta_destino'.
    """
    # me aseguro de que la carpeta destino exista
    os.makedirs(carpeta_destino, exist_ok=True)

    # abro el zip y extraigo todo dentro de esa carpeta
    with zipfile.ZipFile(ruta_zip, 'r') as zf:
        zf.extractall(carpeta_destino)

    # imprimo un mensaje corto para comprobar que ha ido bien
    print(f"he descomprimido '{ruta_zip}' en '{carpeta_destino}'")
    print(f"contenido extraído: {', '.join(sorted(os.listdir(carpeta_destino))[:10])} ...")


def tamanio_archivos_mb(ruta_carpeta: str, nombre1: str, nombre2: str) -> Tuple[float, float]:
    """
    calculo el tamaño en MB de dos archivos ya extraídos.

    Parámetros

    ruta_carpeta : str
        carpeta donde están los ficheros extraídos.
    nombre1 : str
        nombre del primer archivo (con o sin extensión).
    nombre2 : str
        nombre del segundo archivo (con o sin extensión).

    Returns

    (float, float)
        tamaños de los dos archivos en megabytes, con decimales.
    """
    # intento resolver aquí el nombre sin extensión
    def _resolver_path(base: str) -> str:
        # si ya es un archivo, lo devuelvo tal cual
        p = os.path.join(ruta_carpeta, base)
        if os.path.isfile(p):
            return p
        # pruebo con .txt
        p_txt = os.path.join(ruta_carpeta, base + ".txt")
        if os.path.isfile(p_txt):
            return p_txt
        # si no, busco case-insensitive por si hay mayúsculas/minúsculas raras
        base_lower = base.lower()
        for f in os.listdir(ruta_carpeta):
            if f.lower() == base_lower or f.lower() == base_lower + ".txt":
                return os.path.join(ruta_carpeta, f)
        raise FileNotFoundError(f"no encuentro el fichero '{base}' en '{ruta_carpeta}'")

    ruta1 = _resolver_path(nombre1)
    ruta2 = _resolver_path(nombre2)

    # paso bytes a MiB, devuelvo el float
    def _mb(path: str) -> float:
        return os.path.getsize(path) / (1024.0 * 1024.0)

    mb1 = _mb(ruta1)
    mb2 = _mb(ruta2)

    # imprimo el resultado
    print(f"tamaño de '{os.path.basename(ruta1)}': {mb1:.2f} MB")
    print(f"tamaño de '{os.path.basename(ruta2)}': {mb2:.2f} MB")

    return mb1, mb2

# localizo el zip en rutas típicas
candidatos_zip = [
    "data/books.zip",
    "./data/books.zip",
    "Data/books.zip",
    "../data/books.zip",
    "/content/drive/MyDrive/Colab Notebooks/Activity_2/data/books.zip",
    "/content/drive/MyDrive/Programación para la ciencia de datos/Colab Notebooks/Activity_2/data/books.zip",
    "/content/drive/MyDrive/Programacion para la ciencia de datos/Colab Notebooks/Activity_2/data/books.zip",
]
ruta_zip = None
for p in candidatos_zip:
    if os.path.exists(p):
        ruta_zip = p
        break

if ruta_zip is None:
    raise FileNotFoundError("no encuentro 'books.zip'. aseguro que está en Activity_2/data/ con ese nombre exacto.")

# marco la carpeta de salida
carpeta_out = os.path.join(os.path.dirname(ruta_zip), "books_extracted")
zip_decompression(ruta_zip, carpeta_out)

# calculo los tamaños; en mi caso uso 'dracula' y 'sleepyhollow'
_ = tamanio_archivos_mb(carpeta_out, "dracula", "sleepyhollow")
```

En este ejercicio te pedimos que crees una función `process_book` que reciba como entrada la ruta de un archivo de texto (formato .txt) y una expresión regular (patrón) de Python, y devuelva una lista con las **frases** del archivo que contengan alguna coincidencia con el patrón.

En concreto, te pedimos que encuentres:

- Todas las frases del libro **"Drácula"** que contengan la palabra **Dracula** o la palabra **Castle**.  
- Todas las frases del libro **"SleepyHollow"** que contengan la palabra **sleep**.

**Notas importantes:**

1) En ambos casos, los patrones no deben tener en cuenta mayúsculas y minúsculas.  
Es decir, si la misma palabra que tienes que buscar aparece en mayúsculas o minúsculas, debes mostrar ambas.  
Además, solo debes buscar la palabra exacta (por ejemplo, si aparece *Castle* será válido, pero no *Newcastle*, ya que es una palabra diferente aunque contenga “castle”).

2) Como parte del ejercicio, el texto debe separarse en frases. Este es un tema que puede llegar a ser muy complejo, por lo que en este ejercicio trabajaremos con una simplificación.  
En concreto, consideraremos que las frases terminan **siempre** con un punto “.” o con un signo de interrogación “?”.  
Además, el texto ha sido simplificado para no contener algunos símbolos conflictivos como los puntos suspensivos.  
Ten en cuenta que muchas frases están repartidas en más de una línea de texto.  
Las frases que se devuelvan deben incluir, en todo caso, el símbolo de final de frase (punto “.” o “?”).

3) Utiliza los principios de **programación funcional** que hemos visto en teoría para resolver este ejercicio y **expresiones regulares**.

```python
# uso \b en la regex para que sea palabra exacta y separo frases por '.' o '?' manteniendo el signo final
from typing import List
import os
import re

# leo .txt intentando varios encodings por si hay acentos raros
def _leer_txt_robusto(ruta: str) -> str:
    # pruebo encodings típicos; si fallan todos, leo reemplazando caracteres problemáticos
    for enc in ("utf-8", "utf-8-sig", "latin-1", "cp1252"):
        try:
            with open(ruta, "r", encoding=enc) as f:
                return f.read()
        except Exception:
            pass
    with open(ruta, "r", encoding="latin-1", errors="replace") as f:
        return f.read()

# separo el texto en frases simples que acaban en '.' o '?'
def _split_frases_simple(texto: str) -> List[str]:
    # junto saltos de línea en espacios para no romper frases partidas
    txt = re.sub(r"\s+", " ", texto.strip())
    # cojo cualquier cosa hasta encontrar '.' o '?' e incluyo ese signo
    patron = re.compile(r"[^.?]*[.?]")
    frases = [m.group(0).strip() for m in patron.finditer(txt)]
    return frases

def process_book(ruta_txt: str, patron_str: str) -> List[str]:
    """
    leo un libro .txt y devuelvo las frases que cumplen el patrón de la regex.

    parámetros

    ruta_txt : str
        ruta del .txt.
    patron_str : str
        expresión regular en texto.

    returns

    List[str]
        lista de frases (con el '.' o '?' final) que tienen al menos una coincidencia.
    """
    # leo el contenido del archivo
    texto = _leer_txt_robusto(ruta_txt)

    # saco las frases usando la regla simple de arriba
    frases = _split_frases_simple(texto)

    # compilo la regex en modo case-insensitive
    patron = re.compile(patron_str, flags=re.IGNORECASE)

    # limpio espacios y me quedo con las frases que hacen match con la regex
    frases_limpias = map(lambda s: s.strip(), frases)
    frases_match = list(filter(lambda s: bool(patron.search(s)), frases_limpias))

    return frases_match

# busco la carpeta donde dejé los .txt extraídos en el ej.5
candidatos_base = [
    "data/books_extracted",
    "./data/books_extracted",
    "Data/books_extracted",
    "../data/books_extracted",
    "/content/drive/MyDrive/Colab Notebooks/Activity_2/data/books_extracted",
    "/content/drive/MyDrive/Programación para la ciencia de datos/Colab Notebooks/Activity_2/data/books_extracted",
    "/content/drive/MyDrive/Programacion para la ciencia de datos/Colab Notebooks/Activity_2/data/books_extracted",
]
base_dir = None
for p in candidatos_base:
    if os.path.isdir(p):
        base_dir = p
        break
if base_dir is None:
    raise FileNotFoundError("no encuentro 'books_extracted'. antes ejecuto el ejercicio 5.")

# localizo los dos .txt (Dracula y Sleepy Hollow)
ruta_dracula = None
ruta_sleepy = None
for f in os.listdir(base_dir):
    fl = f.lower()
    if "dracula" in fl and f.endswith(".txt"):
        ruta_dracula = os.path.join(base_dir, f)
    if "sleepy" in fl and f.endswith(".txt"):
        ruta_sleepy = os.path.join(base_dir, f)

if ruta_dracula is None:
    raise FileNotFoundError("no encuentro el .txt de drácula en 'books_extracted'.")
if ruta_sleepy is None:
    raise FileNotFoundError("no encuentro el .txt de sleepy hollow en 'books_extracted'.")

# patrones con palabra exacta y sin diferenciar mayúsculas/minúsculas
patron_dracula = r"\b(dracula|castle)\b"
patron_sleep   = r"\bsleep\b"

# ejecuto la función y enseño el recuento + las 5 primeras frases como ejemplo
frases_dracula = process_book(ruta_dracula, patron_dracula)
frases_sleep   = process_book(ruta_sleepy, patron_sleep)

print("— 6.A: Drácula — frases con 'Dracula' o 'Castle' (palabra exacta, case-insensitive) —")
print(f"total de frases encontradas: {len(frases_dracula)}")
for i, s in enumerate(frases_dracula[:5], start=1):
    print(f"{i}. {s}")

print("\n— 6.B: Sleepy Hollow — frases con 'sleep' (palabra exacta, case-insensitive) —")
print(f"total de frases encontradas: {len(frases_sleep)}")
for i, s in enumerate(frases_sleep[:5], start=1):
    print(f"{i}. {s}")
```
